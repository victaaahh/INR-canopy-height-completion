```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
data_dir = "data_var_results/"
```

```python
means = np.load(data_dir + "means.npy")
mean_squares = np.load(data_dir + "mean_squares.npy")
mean_squared_differences = np.load(data_dir + "mean_squared_differences.npy")
```

```python
# Mean
means.mean()
```

```python
# Variance
mean_squared_differences.mean()
```

```python
# Standard Deviation
np.sqrt(mean_squared_differences.mean())
```

```python
# Alternative variance computation
np.mean(mean_squares) - np.mean(means)**2
```

```python
# Alternative standard deviation
np.sqrt(np.mean(mean_squares) - np.mean(means)**2)
```

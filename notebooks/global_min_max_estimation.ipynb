```python
import sys
sys.path.append("..")

from tqdm import tqdm

from src.data import OpenCanopy
import numpy as np
```

```python
opencanopy = OpenCanopy("../datasets/canopy_height", split="train")
```

```python
sample = np.random.choice(range(len(opencanopy)), replace=False, size=1000)
```

```python
global_max = 0
global_min = float("inf")
for i in tqdm(sample):
    _, target = opencanopy[i]
    local_max = target.max()
    if local_max > global_max:
        global_max = local_max

    local_min = target.min()
    if local_min < global_min:
        global_min = local_min

print("Global max:", global_max)
print("Global min:", global_min)
```

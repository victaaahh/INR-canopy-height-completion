```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
data_dir = "data_dist_results/"
```

```python
global_hist = np.load(data_dir + "global_hist.npy")
max_heights = np.load(data_dir + "max_heights.npy")
means = np.load(data_dir + "means.npy")
median_heights = np.load(data_dir + "median_heights.npy")
pixels_larger_than_600 = np.load(data_dir + "pixels_larger_than_600.npy")
pixels_larger_than_6000 = np.load(data_dir + "pixels_larger_than_6000.npy")
vars_ = np.load(data_dir + "vars_.npy")
```

```python
total = np.sum(global_hist)
halfway = total / 2
np.cumsum(global_hist) > halfway
```

```python
percentage_over_600 = pixels_larger_than_600/np.sum(global_hist)*100

_, bins = np.histogram([1,2], bins=6000, range=(0, 6000))

fig, axes = plt.subplots(1, 1, figsize=(5, 2.5), constrained_layout=True)
axes.bar(list(bins[:600] / 10) + [69.9], list(global_hist[:600]) + [pixels_larger_than_600], width=0.1, edgecolor="#5faf66", align="edge", color="#5faf66")
axes.set_yscale("log")
axes.set_ylim(1, 10**10)
axes.set_xlim(-2, 76)
#axes.yaxis.get_major_locator().set_params(numticks=10)
#axes.yaxis.get_minor_locator().set_params(numticks=15, subs=[.2,.6])
axes.set_xlabel("Height [m]")
axes.set_ylabel("Amount of pixels")


axes.set_xticks([0, 10, 20, 30, 40, 50, 60, 70])
axes.set_xticklabels([0, 10, 20, 30, 40, 50, 60, "Above 60"])

axes.text(70, pixels_larger_than_600+100000, f"{percentage_over_600:.4f}%", ha="center")

plt.savefig("figures/pixel_distribution.pgf")
```

```python
# mean and std after filtering and without outliers
```

```python
all_hist_filtered = all_hist[~((all_hist[:,0] / np.sum(all_hist, axis=1)) > 0.99)]
```

```python
# mean no outliers
mean_no_outliers = np.sum(all_hist_filtered[:,:601] * np.arange(0, 601)) / np.sum(all_hist_filtered[:,:601])
mean_no_outliers
```

```python
# std no outliers
mean_squares_no_outliers = np.sum(all_hist_filtered[:,:601] * (np.arange(0, 601)**2)) / np.sum(all_hist_filtered[:,:601])
np.sqrt(mean_squares_no_outliers - mean_no_outliers**2)
```

```python
# Considering cutoff parameter for ratio of 0s
```

```python
counts = np.zeros(101)
for i, x in enumerate(np.arange(0, 1.01, 0.01)):
    counts[i] = np.sum((all_hist[:,0] / np.sum(all_hist, axis=1)) > x)
```

```python
plt.plot(np.arange(0, 1.01, 0.01), counts / 10000, ".")
plt.plot([0.0, 1.0], [1.0, 0.0])
plt.xlabel("x")
plt.ylabel("Ratio of tiles with ratio of 0 > x")
#plt.axvline(0.99)

# 100% of tiles have more than 0% zeros
# 50% of tiles have more than 45% zeros
# 0% of tiles have more than 100% zeros (can impossibly be higher)
```
